```python
x=df.drop("HeartDisease",axis=1)
y=df["HeartDisease"]
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print("training data",x_train.shape)
print("testing data",x_test.shape)
print("---------------------------")
print("training target",y_train.shape)
print("testing target",y_test.shape)
```

```python

num_cols = x_train.select_dtypes(include='number').columns

cat_cols = x_train.select_dtypes(include='object').columns

print("Numerical columns:", list(num_cols))
print("Categorical columns:", list(cat_cols))

```

```python

print("Training data missing values:")
print(x_train.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer
```

```python
num_imputer = SimpleImputer(strategy='median')

x_train[num_cols] = num_imputer.fit_transform(x_train[num_cols])

x_test[num_cols] = num_imputer.transform(x_test[num_cols])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

```python
x_train_encoded = encoder.fit_transform(x_train[cat_cols])
```

```python
x_test_encoded = encoder.transform(x_test[cat_cols])
```

```python
x_train_encoded = pd.DataFrame(
    x_train_encoded,
    columns=encoder.get_feature_names_out(cat_cols),
    index=x_train.index
)

x_test_encoded = pd.DataFrame(
    x_test_encoded,
    columns=encoder.get_feature_names_out(cat_cols),
    index=x_test.index
)
```

```python
print(x_train_encoded.head())
print(x_train_encoded.shape)
```

```python
x_train_final = pd.concat(
    [x_train[num_cols], x_train_encoded],
    axis=1
)

x_test_final = pd.concat(
    [x_test[num_cols], x_test_encoded],
    axis=1
)
```

```python
print(x_train_final.shape)
print(x_test_final.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(
    x_train_final[num_cols]
)
```

```python
x_test_scaled = scaler.transform(
    x_test_final[num_cols]
)
```

```python
x_train_scaled = pd.DataFrame(
    x_train_scaled,
    columns=num_cols,
    index=x_train_final.index
)

x_test_scaled = pd.DataFrame(
    x_test_scaled,
    columns=num_cols,
    index=x_test_final.index
)
```

```python
x_train_final = pd.concat(
    [x_train_scaled, x_train_encoded],
    axis=1
)

x_test_final = pd.concat(
    [x_test_scaled, x_test_encoded],
    axis=1
)
```

```python
print(x_train_final.shape)
print(x_test_final.shape)
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(
    x_train_final[num_cols]
)
```

```python
x_train_scaled = pd.DataFrame(
    x_train_scaled,
    columns=num_cols,
    index=x_train_final.index
)

x_test_scaled = pd.DataFrame(
    x_test_scaled,
    columns=num_cols,
    index=x_test_final.index
)
```

```python
x_train_final = pd.concat(
    [x_train_scaled, x_train_encoded],
    axis=1
)

x_test_final = pd.concat(
    [x_test_scaled, x_test_encoded],
    axis=1
)
```

```python
print(x_train_final.shape)
print(x_test_final.shape)
```

Train Classification Models

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(max_iter=1000)

model_lr.fit(x_train_final, y_train)

y_pred_lr = model_lr.predict(x_test_final)

print("Accuracy:", model_lr.score(x_test_final, y_test))
```

```python
y_prob_lr = model_lr.predict_proba(x_test_final)

print(y_prob_lr[:5])
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=5)

model_knn.fit(x_train_final, y_train)

y_pred_knn = model_knn.predict(x_test_final)

print("Accuracy:", model_knn.score(x_test_final, y_test))
```

```python
from sklearn.svm import SVC

model_svm = SVC()

model_svm.fit(x_train_final, y_train)

y_pred_svm = model_svm.predict(x_test_final)

print("Accuracy:", model_svm.score(x_test_final, y_test))
```

```python
from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()

model_nb.fit(x_train_final, y_train)

y_pred_nb = model_nb.predict(x_test_final)

print("Accuracy:", model_nb.score(x_test_final, y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier

model_dt = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

model_dt.fit(x_train_final, y_train)

y_pred_dt = model_dt.predict(x_test_final)

print("Accuracy:", model_dt.score(x_test_final, y_test))
```

Model Evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
print(classification_report(y_test, y_pred_lr))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred_lr)

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Logistic Regression Confusion Matrix')

plt.show()
```

```python
models = {
    'Logistic Regression': y_pred_lr,
    'KNN': y_pred_knn,
    'SVM': y_pred_svm,
    'Naive Bayes': y_pred_nb,
    'Decision Tree': y_pred_dt
}

for name, predictions in models.items():

    print("\n", name)

    print(classification_report(
        y_test,
        predictions,
        zero_division=0
    ))
```

```python
results = []

for name, predictions in models.items():

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions, zero_division=0),
        'Recall': recall_score(y_test, predictions, zero_division=0),
        'F1 Score': f1_score(y_test, predictions, zero_division=0)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='F1 Score',
    ascending=False
)

results_df
```

```python
results_df.plot(
    x='Model',
    y='Accuracy',
    kind='bar',
    legend=False,
    title='Model Accuracy Comparison'
)

plt.ylabel('Accuracy')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=results_df,
    x='Model',
    y='Accuracy'
)

plt.title('Model Accuracy Comparison')
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig('figures/model_comparison.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=results_df,
    x='Model',
    y='Accuracy'
)

plt.title('Model Accuracy Comparison')
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig('figures/model_comparison.png', dpi=300)
plt.show()
```

```python
best_model_name = results_df.iloc[0]['Model']

best_predictions = models[best_model_name]

cm = confusion_matrix(y_test, best_predictions)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No Disease', 'Heart Disease'],
    yticklabels=['No Disease', 'Heart Disease']
)

plt.title(f'{best_model_name} Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()

plt.savefig('figures/best_model_confusion_matrix.png', dpi=300)
plt.show()
```
